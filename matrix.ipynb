```python
def print_matrix(matrix, step=None):
    """打印矩阵，带有可选的步骤说明"""
    if step:
        print(f"\n步骤 {step}:")
    for row in matrix:
        print("[", end=" ")
        for num in row:
            # 格式化输出，保持整洁
            print(f"{num:8.4f}", end=" ")
        print("]")
    print()

def find_pivot_row(matrix, col, row):
    """找到主元行（当前列中绝对值最大的元素所在的行）"""
    max_row = row
    max_val = abs(matrix[row][col])
    
    for i in range(row, len(matrix)):
        if abs(matrix[i][col]) > max_val:
            max_val = abs(matrix[i][col])
            max_row = i
    return max_row

def gaussian_elimination(matrix):
    """
    高斯消元法求解线性方程组
    输入: 增广矩阵 (每行最后一个元素是常数项)
    输出: 行最简形矩阵和每一步的说明
    """
    # 将矩阵转换为浮点数以避免整数除法问题
    matrix = [list(map(float, row)) for row in matrix]
    
    n_rows = len(matrix)
    n_cols = len(matrix[0])
    step = 1
    
    print("初始矩阵:")
    print_matrix(matrix)
    
    for col in range(n_cols - 1):  # 最后一列是常数项，不需要处理
        # 找到主元行
        pivot_row = find_pivot_row(matrix, col, col)
        
        # 如果主元行不是当前行，则交换行
        if pivot_row != col:
            matrix[col], matrix[pivot_row] = matrix[pivot_row], matrix[col]
            print_matrix(matrix, f"{step}: 交换行 {col+1} 和行 {pivot_row+1} 以获取主元")
            step += 1
        
        # 如果主元接近0，则无法继续消元（矩阵奇异）
        if abs(matrix[col][col]) < 1e-10:
            print("矩阵是奇异的，没有唯一解")
            return matrix
        
        # 将主元行标准化，使主元为1（使用浮点数除法）
        pivot = matrix[col][col]
        if not abs(pivot - 1.0) < 1e-10:  # 如果主元不是1
            for j in range(col, n_cols):
                matrix[col][j] /= pivot  # 使用浮点数除法
            print_matrix(matrix, f"{step}: 将行 {col+1} 除以 {pivot:.4f} 使主元为1")
            step += 1
        
        # 消去下方和上方的元素
        for i in range(n_rows):
            if i != col and not abs(matrix[i][col]) < 1e-10:
                factor = matrix[i][col]
                for j in range(col, n_cols):
                    matrix[i][j] -= factor * matrix[col][j]
                print_matrix(matrix, f"{step}: 行 {i+1} 减去 {factor:.4f} 倍行 {col+1} 以消去第 {col+1} 列元素")
                step += 1
    
    print("行最简形矩阵:")
    print_matrix(matrix)
    
    # 提取解
    solution = [row[-1] for row in matrix]
    print("解为:")
    for i, val in enumerate(solution):
        print(f"x{i+1} = {val:.4f}")
    
    return matrix

```

```python
# 示例使用
if __name__ == "__main__":
    # 示例1: 3x3 方程组
    # 2x + y - z = 8
    # -3x - y + 2z = -11
    # -2x + y + 2z = -3
    matrix1 = [
        [2, 1, -1, 8],
        [-3, -1, 2, -11],
        [-2, 1, 2, -3]
    ]
    
    print("=== 示例 1 ===")
    gaussian_elimination(matrix1)
    
    # 示例2: 2x2 方程组
    # 3x + 2y = 7
    # x - y = 1
    print("\n\n=== 示例 2 ===")
    matrix2 = [
        [3, 2, 7],
        [1, -1, 1]
    ]
    gaussian_elimination(matrix2)
```
