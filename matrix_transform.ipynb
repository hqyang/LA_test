```python
# 线性变换演示：剪切、缩放与旋转

## 代码块1：导入必要的库import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon
import numpy as np

# 设置中文显示
plt.rcParams["font.family"] = ["Times"]
plt.rcParams["axes.unicode_minus"] = False  # 正确显示负号
```

```python
## 代码块2：定义变换函数
def apply_transformation(points, matrix):
    """将变换矩阵应用到点集上"""
    # 每个点是列向量，因此需要转置后相乘，再转置回来
    return np.dot(matrix, points.T).T

def plot_transformation(original_points, transformed_points, title, xlim, ylim):
    """绘制原始图形和变换后的图形，使用固定的坐标范围"""
    fig, ax = plt.subplots(1, 2, figsize=(12, 6))
    
    # 绘制原始图形
    original_polygon = Polygon(original_points, fill=True, alpha=0.5, color='blue', label='原始图形')
    ax[0].add_patch(original_polygon)
    ax[0].scatter(original_points[:, 0], original_points[:, 1], color='blue')
    ax[0].set_title('Original Shape')
    ax[0].grid(True)
    ax[0].set_xlim(xlim)  # 固定x轴范围
    ax[0].set_ylim(ylim)  # 固定y轴范围
    ax[0].set_aspect('equal')  # 保持纵横比
    # ax[0].legend()
    
    # 绘制变换后的图形
    transformed_polygon = Polygon(transformed_points, fill=True, alpha=0.5, color='red', label='变换后图形')
    ax[1].add_patch(transformed_polygon)
    ax[1].scatter(transformed_points[:, 0], transformed_points[:, 1], color='red')
    ax[1].set_title(title)
    ax[1].grid(True)
    ax[1].set_xlim(xlim)  # 与左侧相同的x轴范围
    ax[1].set_ylim(ylim)  # 与左侧相同的y轴范围
    ax[1].set_aspect('equal')  # 保持纵横比
    # ax[1].legend()
    
    plt.tight_layout()
    plt.show()
```

```python
## 代码块3：定义原始图形# 定义一个正方形的顶点作为演示图形
points = np.array([
    [0, 0],  # 原点
    [1, 0],  # 右
    [1, 1],  # 右上
    [0, 1],  # 上
    [0, 0]   # 回到原点，闭合图形
])

# 定义固定的坐标范围（根据所有变换可能达到的范围提前设定）
xlim = [-2, 4]  # x轴范围
ylim = [-2, 4]  # y轴范围
```

```python
# 显示原始图形
plt.figure(figsize=(6, 6))
polygon = Polygon(points, fill=True, alpha=0.5, color='blue')
plt.gca().add_patch(polygon)
plt.scatter(points[:, 0], points[:, 1], color='blue')
plt.title('Original Shape')
plt.grid(True)
plt.xlim(xlim)
plt.ylim(ylim)
plt.gca().set_aspect('equal')
plt.show()
```

```python
## 代码块4：剪切变换 (Shear)# 水平剪切变换
k = 1.5  # 剪切系数
shear_matrix_h = np.array([
    [1, k],
    [0, 1]
])
sheared_points_h = apply_transformation(points, shear_matrix_h)
plot_transformation(points, sheared_points_h, f'Horizontal shear transformation (k={k})', xlim, ylim)

# 垂直剪切变换
k = 1.0
shear_matrix_v = np.array([
    [1, 0],
    [k, 1]
])
sheared_points_v = apply_transformation(points, shear_matrix_v)
plot_transformation(points, sheared_points_v, f'Vertical shear transformation (k={k})', xlim, ylim)
```

```python
## 代码块5：缩放变换 (Scaling)# 缩放变换
sx, sy = 2.0, 1.5  # x和y方向的缩放因子
scale_matrix = np.array([
    [sx, 0],
    [0, sy]
])
scaled_points = apply_transformation(points, scale_matrix)
plot_transformation(points, scaled_points, f'Scaling Transformation (sx={sx}, sy={sy})', xlim, ylim)

# 均匀缩放（各向同性缩放）
s = 1.8  # 缩放因子
uniform_scale_matrix = np.array([
    [s, 0],
    [0, s]
])
uniform_scaled_points = apply_transformation(points, uniform_scale_matrix)
plot_transformation(points, uniform_scaled_points, f'Uniform Scaling Transformation (s={s})', xlim, ylim)
```

```python
## 代码块6：旋转变换 (Rotation)
# 旋转变换 - 45度
theta = np.pi / 4  # 45度（弧度）
angle_deg = np.degrees(theta)  # 转换为角度
rotation_matrix = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])
rotated_points = apply_transformation(points, rotation_matrix)
plot_transformation(points, rotated_points, f'Rotation Transformation ({angle_deg:.1f})', xlim, ylim)

# 旋转变换 - 90度
theta = np.pi / 2  # 90度（弧度）
angle_deg = np.degrees(theta)  # 转换为角度
rotation_matrix_90 = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]
])
rotated_points_90 = apply_transformation(points, rotation_matrix_90)
plot_transformation(points, rotated_points_90, f'Rotation Transformation ({angle_deg:.1f})', xlim, ylim)
```
